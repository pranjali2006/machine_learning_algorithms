```python
import pandas as pd
from sklearn.datasets import load_diabetes

# Load dataset
diabetes = load_diabetes()

# Convert to DataFrame
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)

# Add target column
df["target"] = diabetes.target

# View dataset
print(df.head())
print(df.shape)
```

```python
df.head()
```

```python
#x,y split
x=df.drop("target",axis=1)
y=df['target']
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
# transfromation
from sklearn.preprocessing import StandardScaler
x_scaler=StandardScaler()
x_train_scale=x_scaler.fit_transform(x_train)
x_test_scale=x_scaler.transform(x_test)
```

```python
# traing the model
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor(n_neighbors=5)# assigning the vlaue of k=5
knn.fit(x_train_scale,y_train)
y_pred=knn.predict(x_test_scale)

```

```python
## regression metrics
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,root_mean_squared_error
print("Mean Absolute Error:",mean_absolute_error(y_test,y_pred))
print("Mean Squared Error:",mean_squared_error(y_test,y_pred))
print("R-squared Score:",r2_score(y_test,y_pred))
print("Root Mean Squared Error:",root_mean_squared_error(y_test,y_pred))
```

# cross vlaidation done to find best vlaue of k

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsRegressor

results = []

for k in range(1, 100):
    knn = KNeighborsRegressor(n_neighbors=k)

    scores = cross_val_score(
        knn,
        x_train_scale,
        y_train,
        cv=5,
        scoring="r2"
    )

    results.append((k, scores.mean()))

# Sort by R² from highest to lowest
results_sorted = sorted(results, key=lambda x: x[1], reverse=True)

# Top 5 K values
print(results_sorted[:5])
```

```python
# traing the model
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor(n_neighbors=18)# assigning the vlaue of k=5
knn.fit(x_train_scale,y_train)
y_pred=knn.predict(x_test_scale)

```

```python
## regression metrics
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,root_mean_squared_error
print("Mean Absolute Error:",mean_absolute_error(y_test,y_pred))
print("Mean Squared Error:",mean_squared_error(y_test,y_pred))
print("R-squared Score:",r2_score(y_test,y_pred))
print("Root Mean Squared Error:",root_mean_squared_error(y_test,y_pred))
```
