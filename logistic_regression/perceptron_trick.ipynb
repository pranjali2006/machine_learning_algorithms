```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                        n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.show()
```

```python
### creating a perceptron function

#for x we add (1) before that all input columns its used for bias 
# creating a array called weigths which will of the size of the columns x has which is 3 columns
# creating a learning rate 
# now creating a loop which will learn 1000 times
# first it will select a random number(i.e student)
#y_pred varivble where we will do the dot product and check is it postivie or negative(means wheather the placement will be ter or not)
# now update the weights as per the formula
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
    return weights[0],weights[1:]
        
```

```python
## creating a function to ckeck if the student is =0 or not

def step(z):
    return 1 if z > 0 else 0
```

```python
intercept_,coef_=perceptron(X,y)
```

```python
print(coef_)
```

```python
print(intercept_)
```

```python

# converting the weights into slope and intercept of the line
# i.e converting the line equation of ax+by+c=0 into y=mx+b form
# here a is the coef of x and b is the coef of y and c is the intercept
# m= - a/b and b= -c/b



m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
# this is used to draw the decision boundary

x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
## now plotting the line and the data points

## in the very first code if the class_sep is 10 then the line will not be able to separate the data points but if we increase the class_sep to 20 then the line will be able to separate the data points
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

<h3>using logistic regression</h3>

```python
from sklearn.linear_model import LogisticRegression
lor = LogisticRegression()
lor.fit(X,y)
```

```python
# concerting to the original line euation of ax+by+c=0 into y=mx+b form
m = -(lor.coef_[0][0]/lor.coef_[0][1])
b = -(lor.intercept_/lor.coef_[0][1])
```

```python
# getting the line equation of logistic regression
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input + b
```

```python
# the left line is the line of logistic regression and the right line is the line of perceptron 

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.plot(x_input1,y_input1,color='black',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
