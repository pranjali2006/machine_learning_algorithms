```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_circles
```

```python
np.random.seed(42)
x,y=make_circles(n_samples=500,factor=0.1,noise=0.35,random_state=42)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import cross_val_score

ada=AdaBoostClassifier()
np.mean(cross_val_score(ada,x,y,scoring="accuracy",cv=10))
```

```python
# using hperparameters in the adboost algorithm

aa=AdaBoostClassifier(n_estimators=1000,learning_rate=0.1)
aa.fit(x,y)
np.mean(cross_val_score(aa,x,y,scoring="accuracy",cv=10))
```
