```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
```

```python
df = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)

df["Price"] = housing.target
```

```python
x= df.drop("Price", axis=1)

y = df["Price"]
```

```python
from sklearn.linear_model import LinearRegression

from sklearn.tree import DecisionTreeRegressor

from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import cross_val_score
```

```python
c1=LinearRegression()
c2=DecisionTreeRegressor()
c3= KNeighborsRegressor()
```

```python
score=cross_val_score(c1,x,y,cv=10,scoring="r2")
print(np.mean(score))
```

```python
score=cross_val_score(c2,x,y,cv=10,scoring="r2")
print(np.mean(score))
```

```python
score=cross_val_score(c3,x,y,cv=10,scoring="r2")
print(np.mean(score))
```
