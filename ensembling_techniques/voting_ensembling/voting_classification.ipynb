```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("C:\\Users\\Pranjali\\OneDrive\\Desktop\\ml_algorithms\\ensembling_techniques\\voting_ensembling\\Iris.csv")
```

```python
df.head()
```

```python
# removing the id column
df=df.iloc[:,1:]
```

```python
df.head()
```

```python
#converting sepices cloumn 
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
```

```python
df['Species']=encoder.fit_transform(df['Species'])
```

```python
df.head()
```

```python
#seprating x and y i.e input and output
x=df.drop("Species",axis=1)
y=df['Species']
```

```python
x.shape
```

```python
y.shape
```

```python
# importing multiple modes for taining the data
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
```

```python
c1=LogisticRegression(max_iter=1000)
c2=KNeighborsClassifier()
c3=RandomForestClassifier(random_state=42)
```

```python
score=cross_val_score(c1,x,y,cv=10,scoring="accuracy")
print(np.mean(score))
```

```python
score=cross_val_score(c2,x,y,cv=10,scoring="accuracy")
print(np.mean(score))
```

```python
score=cross_val_score(c3,x,y,cv=10,scoring="accuracy")
print(np.mean(score))
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
voting = VotingClassifier(
    estimators=[
        ('lr', c1),
        ('knn', c2),
        ('rf', c3)
    ],
    voting='soft'# used to take the majority outcome
)
```

```python
# checking the accuracy of voting classfier after combining the all 3 algortihms 
score=cross_val_score(voting,x,y,cv=10,scoring="accuracy")
print(np.mean(score))
```
